```python
from sklearn.linear_model import LinearRegression

# Dados de entrada (X: minutos) e de saída (y: preço)[span_2](start_span)[span_2](end_span)
X, y = [[15], [30], [45], [60]], [10.0, 15.0, 20.0, 25.0]

# Treinar o modelo de regressão linear[span_3](start_span)[span_3](end_span)
modelo = LinearRegression().fit(X, y)

# Efetuar previsões para 45, 20 e 60 minutos[span_4](start_span)[span_4](end_span)
for t in [45, 20, 60]:
    print(f"Preço para {t} min: R$ {modelo.predict([[t]])[0]:.2f}")
```
